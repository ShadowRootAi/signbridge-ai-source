# SignBridge AI · Πηγαίος κώδικας

Τρέξε το πρώτο κελί και διάλεξε το `SignBridge-Colab-Source.zip`. Το notebook θα επαληθεύσει και θα ανοίξει το αρχείο στο προσωρινό περιβάλλον του Colab. Ο φάκελος **Files** στο πλάι δείχνει τα αρχεία. Δεν απαιτούνται κλειδιά API ή σύνδεση Google Drive.

Το δεύτερο κελί ελέγχει βασικές συναρτήσεις αξιολόγησης με συνθετικά δεδομένα. Για αναγνώριση από πραγματικό βίντεο, χρησιμοποίησε το [δημόσιο SignBridge](https://signbridge-ai-gsl.pages.dev/#/try).

```python
from google.colab import files
from hashlib import sha256
from io import BytesIO
from pathlib import Path
from tempfile import mkdtemp
from zipfile import ZipFile

name = 'SignBridge-Colab-Source.zip'
uploaded = files.upload()
if set(uploaded) != {name}:
    raise ValueError(f'Επίλεξε μόνο το {name}.')

archive = uploaded[name]
if sha256(archive).hexdigest() != 'fcbfb523cc739739489c70e9740647ab125efbda8f0d8b6f5719044b887da169':
    raise ValueError('Το ZIP διαφέρει από το ελεγμένο αρχείο.')

destination = Path(mkdtemp(prefix='signbridge-', dir=Path.cwd()))
with ZipFile(BytesIO(archive)) as bundle:
    bundle.extractall(destination)

source = destination / 'signbridge-source'
print(f'Ο κώδικας βρίσκεται στον φάκελο: {source}')
print((source / 'COLAB.md').read_text(encoding='utf-8'))
```

```python
import importlib.util
import subprocess
import sys

if importlib.util.find_spec('numpy') is None:
    print('Το μικρό τεστ χρειάζεται NumPy. Ο κώδικας είναι ήδη διαθέσιμος στον φάκελο Files.')
else:
    subprocess.run([sys.executable, '-m', 'unittest', 'discover', '-s', 'ml', '-p', 'test_common.py'],
                   cwd=source, check=True)
```
